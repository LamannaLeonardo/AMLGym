# Offline Algorithms

AMLGym includes several algorithms for offline learning domain models from full/partial/noisy trajectories.

```python
!pip install amlgym
```

```python
from amlgym.benchmarks import get_domain_path, get_trajectories_path
from amlgym.algorithms import print_algorithms, get_algorithm
print_algorithms()
```

## Example 1: offline learning with full observability

Instantiate an algorithm for offline learning domain models in fully observable environments (e.g., [SAM](https://proceedings.kr.org/2021/36/)):

```python
learner = get_algorithm('sam')
```

Get an empty IPC domain:

```python
from amlgym.util.util import empty_domain

domain_path = get_domain_path('blocksworld')

domain_empty_path = empty_domain(domain_path)
with open(domain_empty_path, 'r') as f:
    print(f.read())
```

Get a set of trajectories associated with the domain:

```python
traj_paths = get_trajectories_path('blocksworld')
```

Learn a domain model specified in PDDL:

```python
model = learner.learn(domain_empty_path, traj_paths)
print(model)
```

## Example 2: offline learning with partial observability

Instantiate an algorithm for offline learning domain models in partially observable environments (e.g., [OffLAM](https://www.sciencedirect.com/science/article/abs/pii/S0004370224001929)):

```python
learner = get_algorithm('offlam')
```

Get an empty IPC domain and associated set of trajectories:

```python
domain_path = get_domain_path('blocksworld')
domain_empty_path = empty_domain(domain_path)
traj_paths = get_trajectories_path('blocksworld')
```

Learn a domain model specified in PDDL:

```python
model = learner.learn(domain_empty_path, traj_paths)
print(model)
```

## Example 3: offline learning with noisy observability

Instantiate an algorithm for offline learning domain models in noisy environments (e.g., [NOLAM](https://ojs.aaai.org/index.php/ICAPS/article/view/31493)):

```python
learner = get_algorithm('nolam', noise=0.1)
```

Get an empty IPC domain and associated set of trajectories:

```python
domain_path = get_domain_path('blocksworld')
domain_empty_path = empty_domain(domain_path)
traj_paths = get_trajectories_path('blocksworld')
```

Learn a domain model specified in PDDL:

```python
model = learner.learn(domain_empty_path, traj_paths)
print(model)
```
