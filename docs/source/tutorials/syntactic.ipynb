# AMLGym metrics

AMLGym provides several measures for evaluation an action model w.r.t. a reference model or simulator.

```python
!pip install amlgym
```

```python
from amlgym.metrics import *
print_metrics()
```

Learn a model to be evaluated:

```python
from amlgym.benchmarks import *
from amlgym.algorithms import *

offlam = get_algorithm('OffLAM')
domain_ref_path = get_domain_path('blocksworld')
traj_paths = get_trajectories_path('blocksworld')
model = offlam.learn(domain_ref_path, traj_paths)
print(model)
```

Save the learned model to a file:

```python
domain_eval_path = 'domain_learned.pddl'
with open(domain_eval_path, 'w') as f:
    f.write(model)
```

## Syntactic evaluation

Syntactic similarity can be measured by means of precision and recall:

```python
precision = syntactic_precision(domain_eval_path, domain_ref_path)
print(precision)
```

```python
recall = syntactic_recall(domain_eval_path, domain_ref_path)
print(recall)
```

## Problem solving evaluation

Get the set of problems necessary to evaluate problem solving metrics:

```python
probs_paths = get_problems_path('blocksworld', kind='solving')
print(probs_paths)
```

Solve the problems with the model to be evaluated and evaluate it in a simulator defined by a reference model:

```python
from amlgym.metrics import problem_solving
metrics = problem_solving(domain_eval_path, domain_ref_path, probs_paths, timeout=60)
print(metrics)
```
